## 8. Confusion Matrix interpretieren

Diese Beispielwerte dienen nur zur Interpretation. Es wird noch kein eigenes Modell trainiert.

```python
X = df[
    [
        "duration",
        "amount",
        "installment_rate",
        "age",
        "number_credits"
    ]
]

y = df["credit_risk"]

#Numerische Spalten:
#['duration', 'amount', 'installment_rate','age', 'number_credits'],[ 'people_liable', 'credit_risk','present_residence','number_credits']

#Kategoriale Spalten:
#['status', 'credit_history', 'purpose', 'savings', 'employment_duration', 'personal_status_sex', 'other_debtors', 'property', 'other_installment_plans', 'housing', 'job', 'telephone', 'foreign_worker']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.calibration import CalibratedClassifierCV
from sklearn.svm import SVC

pipeline = Pipeline(
    [
        ("scaler", MinMaxScaler()),
        ("svm", CalibratedClassifierCV(
            SVC(kernel="rbf"),
            ensemble=False
        )),
    ]
)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler

param_grid = {
    "scaler": [
        StandardScaler(),
        MinMaxScaler(),
    ],
    "svm__estimator__kernel": [
        "linear",
        "rbf",
        "poly",
    ],
    "svm__estimator__C": [
        0.1,
        1,
        10,
    ],
    "svm__estimator__gamma": [
        "scale",
        0.01,
        0.001,
    ],
}

grid = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid.fit(X_train, y_train)
```

```python
best_svm_model = grid.best_estimator_

print("Beste Parameter:")
print(grid.best_params_)
```

```python
y_pred = best_svm_model.predict(X_test)
y_proba = best_svm_model.predict_proba(X_test)[:, 1]

print("Vorhergesagte Klassen:")
print(y_pred)

print("\nWahrscheinlichkeiten für Kundenabwanderung:")
print(y_proba)
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("Optimierte Modellgenauigkeit:", accuracy)

print("\nKlassifikationsbericht:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Schlechtes Kreditrisiko",
        "Gutes Kreditrisiko"
    ]
)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix – Kreditrisiko")
plt.show()

print(cm)
```

### Reflexion Confusion Matrix

    Die Klasse 1 wird sehr gut erkannt.
    133 Fälle wurden korrekt als Klasse 1 klassifiziert.

    False Positives sind Fälle, die fälschlich als positiv (Klasse 1) vorhergesagt wurden, obwohl sie tatsächlich Klasse 0 sind.

    False Negatives sind Fälle, die fälschlich als negativ (Klasse 0) vorhergesagt wurden, obwohl sie tatsächlich Klasse 1 sind.

5. Warum reicht Accuracy allein nicht aus?

## 9. ROC-Kurve verstehen

```python

from sklearn.metrics import roc_curve, roc_auc_score

fpr, tpr, thresholds = roc_curve(y_test, y_proba)
auc = roc_auc_score(y_test, y_proba)

plt.figure(figsize=(6, 5))
plt.plot(
    fpr,
    tpr,
    label=f"ROC-Kurve (AUC = {auc:.2f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray",
    label="Zufall"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC-Kurve – Kundenabwanderung")
plt.legend()
plt.grid(True)

plt.show()

print(f"AUC: {auc:.2f}")
```

### Reflexion ROC-Kurve
Reflexion ROC-Kurve
1. Was zeigt die ROC-Kurve?

Die ROC-Kurve (Receiver Operating Characteristic) stellt die Beziehung zwischen der True Positive Rate (TPR) und der False Positive Rate (FPR) für verschiedene Schwellenwerte dar. Sie zeigt, wie gut ein Modell zwischen positiven und negativen Klassen unterscheiden kann.

2. Was bedeutet eine gute Kurve?

Eine gute ROC-Kurve verläuft möglichst nahe an der linken oberen Ecke des Diagramms. Das bedeutet, dass viele positive Fälle korrekt erkannt werden (hohe TPR), während gleichzeitig nur wenige negative Fälle fälschlicherweise als positiv klassifiziert werden (niedrige FPR). Die gestrichelte Diagonale entspricht einem Zufallsmodell.

3. Was bedeutet AUC?

Die AUC (Area Under the Curve) misst die Fläche unter der ROC-Kurve und dient als Kennzahl für die Trennfähigkeit des Modells. Ein Wert von 0,5 entspricht einer zufälligen Vorhersage, während ein Wert von 1,0 eine perfekte Klassifikation bedeutet. Die vorliegende AUC von 0,74 zeigt, dass das Modell die Klassen insgesamt gut voneinander unterscheiden kann.

4. Wann ist ROC hilfreicher als Accuracy?

Die ROC-Kurve ist besonders hilfreich bei unausgeglichenen Datensätzen, bei denen eine hohe Accuracy irreführend sein kann. Außerdem ermöglicht sie die Bewertung des Modells über verschiedene Schwellenwerte hinweg und zeigt den Kompromiss zwischen Sensitivität und Fehlalarmen. Dadurch liefert sie oft aussagekräftigere Informationen als die reine Accuracy.

Fazit:
Mit einer AUC von 0,74 erzielt das Modell eine solide Leistung und unterscheidet die Klassen deutlich besser als ein Zufallsmodell. Es besteht jedoch noch Potenzial zur Verbesserung der Vorhersagequalität.
