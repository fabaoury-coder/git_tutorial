## 3. Wann ist ein Dashboard sinnvoll?

Beantworten Sie kurz:

1. Wann genügt eine Tabelle?

Eine Tabelle genügt, wenn wenige Datenpunkte exakt und detailliert verglichen oder nachgeschlagen werden sollen (z. B. konkrete Werte, Listen, kleine Datensätze).

2. Wann genügt ein einzelnes Diagramm?

Ein einzelnes Diagramm reicht, wenn nur eine zentrale Kennzahl oder ein klarer Zusammenhang visualisiert werden soll (z. B. Trend, Verteilung oder Vergleich einer Variable

3. Wann wird ein Dashboard notwendig?

Ein Dashboard wird notwendig, wenn mehrere Kennzahlen gleichzeitig überwacht und kombiniert dargestellt werden müssen, besonders bei komplexen Daten oder Entscheidungsprozessen in Echtzeit oder regelmäßigem Monitoring.

4. Welche Zielgruppe würde ein Dashboard nutzen?

Dashboards werden vor allem von Entscheidungsträgern und Anwendern mit Analysebedarf genutzt, z. B. Manager, Analysten, Projektleiter oder Fachabteilungen.

### Meine Antwort

Eine Tabelle genügt, wenn Einzelwerte geprüft werden sollen.
Ein Diagramm genügt, wenn eine einzelne Verteilung oder ein Zusammenhang gezeigt werden soll.
Ein Dashboard ist sinnvoll, wenn mehrere Kennzahlen und Diagramme gemeinsam eine Entscheidung unterstützen sollen.

## 7. Zielvariable als interaktives Balkendiagramm

```python
import plotly.express as px

credit_risk_counts = df["credit_risk"].value_counts().reset_index()
credit_risk_counts.columns = ["credit_risk", "Anzahl"]

fig = px.bar(
  credit_risk_counts,
    x="credit_risk",
    y="Anzahl",
    text="Anzahl",
    title="Verteilung der Zielvariable credit_risk",
    labels={
        "credit_risk": "Kreditrisiko",
        "Anzahl": "Anzahl der Kunden"
    }
)

fig.show()
```

### Interpretation

Die Zielvariable ist nicht vollständig ausgeglichen.
Mehr Personen haben nicht überlebt als überlebt.
Für ein späteres Klassifikationsmodell bedeutet das, dass Accuracy allein nicht ausreichen könnte.

## 8. KPI berechnen

```python
anzahl_kunden = len(df)

kredit_risiko_rate = df["credit_risk"].mean()
durchschnittsalter = df["age"].mean()
durchschnitt_kreditbetrag = df["amount"].mean()

print("Anzahl Kunden:", anzahl_kunden)
print("Kreditrisiko-Rate:", round(kredit_risiko_rate * 100, 2), "%")
print("Durchschnittsalter:", round(durchschnittsalter, 2))
print("Durchschnittlicher Kreditbetrag:", round(durchschnitt_kreditbetrag, 2))
```

## 9. KPI-Karte mit Plotly

```python
import plotly.graph_objects as go

kredit_rate = df["credit_risk"].mean() * 100

fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number",
    value=kredit_rate,
    title={"text": "Anteil credit_risk = 1 in %"},
    number={"suffix": "%", "valueformat": ".2f"}
))

fig.update_layout(
    title="KPI: Kreditqualität (German Credit)",
    height=300
)

fig.show()
```

### Interpretation KPI

Die KPI verdichtet die Zielvariable auf einen zentralen Wert.
Sie eignet sich für ein Dashboard, weil sie schnell einen Überblick gibt.
Allein reicht sie aber nicht aus, da Ursachen und Gruppenunterschiede noch nicht sichtbar sind.

## 10. Dashboard-Elemente vorbereiten

```python

import plotly.express as px
fig1 = px.histogram(
    df,
    x="age",
    color="credit_risk",
    title="Altersverteilung nach Kreditrisiko"
)

fig2 = px.box(
    df,
    x="credit_risk",
    y="amount",
    title="Kreditbetrag nach Kreditrisiko"
)

fig3 = px.bar(
    df.groupby("job", observed=True)["credit_risk"]
      .mean()
      .reset_index(),
    x="job",
    y="credit_risk",
    title="Kreditrisiko nach Beruf",
    labels={
        "credit_risk": "Durchschnittliches Kreditrisiko",
        "job": "Beruf"
    }
)

fig1.show()
fig2.show()
fig3.show()
#Chaque colonne = une feature
```
