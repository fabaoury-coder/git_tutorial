## 7. Python-Beispiel: GermanCredit-Pipeline

Die folgenden Funktionen bilden die Pipeline technisch nach. Jede Funktion entspricht einem Verarbeitungsschritt.

````python
def load_data():
    data = df.copy()
    print("Daten wurden geladen.")
    print("Form:", data.shape)
    return data


def check_data_quality(data):
    print("Datenqualität wird geprüft.")
    print("Zeilen und Spalten:", data.shape)
    print("\nFehlende Werte:")
    print(data.isna().sum())
    print("\nDubletten:", data.duplicated().sum())
    print("\nDatentypen:")
    print(data.dtypes)
    return data


def clean_data(data):
    print("Daten werden bereinigt.")
    data = data.drop_duplicates()
    return data


def create_features(data):
    print("Features und Zielvariable werden erstellt.")
    X = data.drop(columns=["credit_risk"])
    y = data["credit_risk"]
    return X, y


def split_data(X, y):
    print("Trainings- und Testdaten werden erstellt.")
    return train_test_split(
        X,
        y,
        test_size=0.25,
        random_state=42,
        stratify=y
    )


def train_model(X_train, y_train):
    print("Modell wird trainiert.")

    numeric_features = ["duration", "amount", "installment_rate", "present_residence", "age", "number_credits", "people_liable"]
    categorical_features = ["status", "credit_history"]

    numeric_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer(transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ])

    model = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(random_state=42))
    ])

    model.fit(X_train, y_train)
    return model


def evaluate_model(model, X_test, y_test):
    print("Modell wird bewertet.")
    y_pred = model.predict(X_test)

    metrics = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0)
    }

    cm = confusion_matrix(y_test, y_pred)
    return metrics, cm
````

````python
data = load_data()
data = check_data_quality(data)
data = clean_data(data)
data = check_data_quality(data)

X, y = create_features(data)
X_train, X_test, y_train, y_test = split_data(X, y)

model = train_model(X_train, y_train)
metrics, cm = evaluate_model(model, X_test, y_test)

metrics
````

## 10a. Optional: Automatisiertes Labeling mit Snorkel

### Theorie

In manchen Machine-Learning-Projekten existieren keine fertigen Labels. Das manuelle Kennzeichnen großer Datenmengen ist zeitaufwendig und teuer.

Snorkel verfolgt den Ansatz der sogenannten Weak Supervision. Statt jeden Datensatz manuell zu labeln, werden Regeln (Labeling Functions) definiert. Diese erzeugen automatisch vorläufige Labels.

Mehrere Regeln können anschließend kombiniert werden, um eine gemeinsame Labelentscheidung zu treffen.

Typische Anwendungsfälle:

- Kundenfeedback klassifizieren
- Support-Tickets priorisieren
- E-Mails vorsortieren
- Dokumente kategorisieren

Wichtig: Die erzeugten Labels müssen anschließend überprüft werden.

### Beispielhafter Datensatz

````python
feedback = pd.DataFrame({
    "text": [
        "Der Service war sehr gut.",
        "Die Lieferung war langsam.",
        "Ich habe eine Rechnung erhalten.",
        "Das Produkt ist empfehlenswert."
    ]
})

feedback
````

### Snorkel installieren

````python
from snorkel.labeling import labeling_function

ABSTAIN = -1
NEGATIV = 0
NEUTRAL = 1
POSITIV = 2


@labeling_function()
def lf_positiv(x):

    text = x.text.lower()

    if "gut" in text or "empfehlenswert" in text:
        return POSITIV

    return ABSTAIN


@labeling_function()
def lf_negativ(x):

    text = x.text.lower()

    if "langsam" in text:
        return NEGATIV

    return ABSTAIN


@labeling_function()
def lf_neutral(x):

    text = x.text.lower()

    if "rechnung" in text:
        return NEUTRAL

    return ABSTAIN
````

````python
from snorkel.labeling import PandasLFApplier

lfs = [
    lf_positiv,
    lf_negativ,
    lf_neutral
]

applier = PandasLFApplier(lfs=lfs)

L_train = applier.apply(df=feedback)

L_train
````

````python
from snorkel.labeling.model import LabelModel

label_model = LabelModel(
    cardinality=3,
    verbose=False
)

label_model.fit(
    L_train=L_train,
    n_epochs=500
)

feedback["label"] = label_model.predict(L=L_train)

feedback
````

Interpretation

Dieses Beispiel zeigt den Unterschied zwischen einer einzelnen Regel und einem Snorkel-Workflow. Mehrere Labeling Functions werden kombiniert und liefern gemeinsam ein vorläufiges Label.

Für die Customer-Churn-Pipeline ist Snorkel normalerweise nicht notwendig, da die Zielvariable churn bereits vorhanden ist.

12a. Erweiterte Qualitätskontrolle mit Pandera

Theorie

Professionelle Data Pipelines prüfen Daten häufig gegen ein fest definiertes Schema.

Dadurch können Fehler bereits vor dem Modelltraining erkannt werden.

Pandera ermöglicht unter anderem die Prüfung von:

* Datentypen
* Wertebereichen
* Pflichtfeldern
* Kategorien
* Spaltenstrukturen

Wenn eine Validierung fehlschlägt, sollte die Pipeline gestoppt werden.

Pandera installieren

Bibliothek importieren

````python
import pandera as pa
from pandera import Column
from pandera import DataFrameSchema
from pandera import Check
````

Schema definieren

````python

credit_risk_schema = DataFrameSchema({

    "duration": Column(int,
                       Check.ge(1)
                       ),

    "amount": Column(int,
                     Check.ge(0)
                     ),

    "age": Column(int,
                  Check.ge(0)
                  ),

    "present_residence": Column(int,
                                Check.ge(0)
                                ),

    "status": Column(
        str,
        Check.isin([
            "... < 100 DM",
            "0 <= ... < 200 DM",
            "... >= 200 DM / salary for at least 1 year",
            "no checking account"
        ])
    ),

    "credit_risk": Column(
        int,
        Check.isin([0, 1])
    )
})

validierte_daten = credit_risk_schema.validate(df)

print("Validierung erfolgreich.")
````

````python
alidierte_daten = credit_risk_schema.validate(df)
print("Validierung erfolgreich.")
````

Interpretation

Die Pandera-Prüfung erkennt:

* fehlende Spalten
* falsche Datentypen
* ungültige Kategorien
* unzulässige Werte

In produktiven Data Pipelines wird die Schema-Validierung häufig direkt nach dem Laden der Daten durchgeführt.

Pandera ergänzt klassische Prüfungen wie Missing-Value-Checks oder Dublettenprüfungen und erhöht die Zuverlässigkeit der Pipeline.
